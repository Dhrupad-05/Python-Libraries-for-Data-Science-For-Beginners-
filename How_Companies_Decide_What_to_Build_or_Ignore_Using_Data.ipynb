```python
import pandas as pd
df= pd.read_csv('/content/mckinsey (1).csv')
df.sample(10) #shows random values
```

```python
df.info()
```

```python
df['country'].nunique()
```

```python
#avg life expectancy of India
comp_op=df['country'] == 'India' #comparing
df[comp_op]#showing the isolated "True" values of the above comparison
```

```python
df_ind= df[comp_op]
df['life_exp'].mean()
df['life_exp'].min()
df['life_exp'].max()
```

**country with lowest expentancy**

```python

min_exp=df['life_exp'].min() #atoring the miminum life expectancy in a variable
df['life_exp']==min_exp #check in which rows the min value is present... gives a bunch of true/falses
#we can do this in one line rather than sdoing it in several
df[df['life_exp']==min_exp]
```

```python
#Maximum Gdp
max_gdp=df['gdp_cap'].max()
df[df['gdp_cap']==max_gdp]
```

```python
#Minimum gdp
min_gdp=df['gdp_cap'].min()
df[df['gdp_cap']==min_gdp]
```

**GDP of every continent in 2007**

```python
df_2007 = df[df['year']==2007]
df.head()
```

```python
df_2007['gdp']=df_2007['population']*df_2007['gdp_cap'] #new column created
df_2007.head()
```

```python
#rank countries by gdp
df_2007.sort_values(by='gdp')#asending
df_2007.sort_values(by='gdp',ascending=False) #decensing
```

```python
df_2007.groupby(by='continent')['gdp'].sum()
```

```python
df_2007.groupby(by='continent')[['gdp','gdp_cap']].sum()
```

**Average Life Expentency of a Continent**

```python
df_2007.groupby(by='continent')[['life_exp']].mean()
```
